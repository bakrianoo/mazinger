# 🎬 Mazinger Dubber — Colab Quickstart

Run the **Mazinger Studio** web UI on a free Colab T4 GPU.

**Before you start:**

1. Turn the GPU on — `Runtime → Change runtime type → T4 GPU`.

Then run the three cells below in order:

1. **Install** Mazinger and `ffmpeg` (~2–3 min).
2. **Sign in** to Hugging Face (optional — see above).
3. **Launch** the web UI, and open the `https://….gradio.live` link it prints.

> Keep this Colab tab open while you use the UI. Closing it stops the server.

```python
# 1. Install ffmpeg + Mazinger (full pipeline + Studio web UI)
#
# uv is required, not a preference: mazinger[all] bundles packages with
# conflicting dependency pins that plain pip cannot resolve.
!apt-get -qq install -y ffmpeg
!pip install -q uv
!uv pip install -q --system "mazinger[all]"

# YouTube challenges requests from datacenter IPs, which is what Colab is.
# This plugin mints the "proof of origin" tokens that get downloads working;
# yt-dlp picks it up automatically once installed.
!uv pip install -q --system bgutil-ytdlp-pot-provider
```
